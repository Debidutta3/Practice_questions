```python
arr = [10, 5, 8, 20, 15]

max = arr[0]

for i in range(1, len(arr)):
  if arr[i] > max:
    max = arr[i]
arr.remove(max)

second_max = arr[0]
for i in range(1, len(arr)):
  if arr[i] > second_max:
    second_max = arr[i]

print("Second largest element is:", second_max)
```

```python
arr = [1, 2, 3, 3, 5]

freq = {}

for num in arr:
    freq[num] = freq.get(num, 0) + 1

repeated = None

for num, count in freq.items():
    if count > 1:
        repeated = num

n = len(arr)
expected_sum = n * (n + 1) // 2
actual_sum = sum(arr)

missing = expected_sum - (actual_sum - repeated)

print("Missing number:", missing)
print("Repeated number:", repeated)
```

```python
n = 19

prime = True

if n<=1:
  prime = False
else:
  for i in range(2, int(n**0.5)+1):
    if n%i == 0:
      prime =False
      break

if prime:
  print("Prime")
else:
  print("Not Prime")
```

```python
n=7
def fib(n):
  if n <= 0:
    return 0
  elif n == 1:
    return 1
  return fib(n-1) + fib(n-2)

for i in range(7):
  print(fib(i))
```

```python
a = 15
b = 25

gcd = 1
# Use python's built-in min function (or a manual alternative if min is overridden in the namespace)
limit = a if a < b else b

for i in range(1, limit + 1):
  if a % i == 0 and b % i == 0:
    gcd = i

gcd
```

```python
n = 153
temp = n

length = len(str(n))
sum = 0

while n > 0:
  rem = n%10
  sum += rem**length
  n = n//10

if sum == temp:
  print("Armstrong")
else:
  print("Not Armstrong")
```
