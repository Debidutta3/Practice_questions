```python
#count unique pairs

def count_unique(nums, target):
  new_nums = list(set(nums))
  count = 0

  for i in range(len(new_nums)):
    for j in range(i+1, len(new_nums)):

      if new_nums[i] + new_nums[j] == target:
        count+=1

  return count

count_unique([1, 5, 7, -1, 5], 6)
```

```python
# Best Time to Buy and Sell Stock

def buy_sell(prices):

  minimum = min(prices)

  for i in range(len(prices)):
    if prices[i] == minimum and i == len(prices)-1:
      return 0

  max_val = float('-inf')

  for i in range(len(prices)):
    for j in range(i+1, len(prices)):
      max_val = max(max_val, prices[j]-prices[i])

  return max_val

buy_sell([7,1,5,3,6,4])
```

```python
#majority element

def majority_element(nums):
  freq = {}

  for num in nums:
    freq[num] = freq.get(num, 0) + 1

  maxi = float('-inf')

  for num, count in freq.items():
    if count > maxi:
      maxi = count
      most_repeated = num

  return most_repeated

majority_element([2,2,1,1,1,2,2])
```

```python
# Search in Rotated Sorted Array
def search(nums, target):

    for i in range(len(nums)):
      if nums[i] == target:
        return i

    return -1

search([4, 5, 6, 7, 0, 1, 2], 3)
```

```python
#Merge Two Sorted Arrays

def merge(nums1, nums2):

  l = []
  l = nums1 + nums2

  # for i in l:
  #   if i == 0:
  #     l.remove(i)

  l.sort()
  l1 = []
  for i in l:
    if i != 0:
      l1.append(i)

  return l1

merge([1,2,3,0,0,0], [2,5,6])
```

```python
# 3Sum

def three_sum(nums):
  for i in range(len(nums)):
    for j in range(i+1, len(nums)):
      for k in range(j+1, len(nums)):
        if nums[i] + nums[j] + nums[k] == 0:
          return [i,j,k]
  return None

three_sum([-1,0,1,2,-1,-4])
```
