```python
#Find the first non-repeating character in a string.

def non_repeat(s):
  freq={}

  for ch in s:
    freq[ch] = freq.get(ch, 0) + 1

  # Iterate through the original string to preserve order
  for ch in s:
    if freq[ch] == 1:
      return ch
  return None

print(non_repeat('abcdef'))
print(non_repeat('abcabcdef'))
print(non_repeat('aabbcc'))
```

```python
#Given an array of integers, move all zeros to the end while maintaining the relative order of non-zero elements.

def move_zeroes(nums):

  l = []

  for i in nums:
    if i != 0:
      l.append(i)

  for j in range(len(nums) - len(l)):
    l.append(0)

  return l
print(move_zeroes([0,1,0,3,12]))
```

```python
#find missing number

def missing_num(nums):

  n = len(nums) + 1
  expected_sum = n * (n+1) // 2
  sum_num = sum(nums)

  return expected_sum - sum_num
print(missing_num([1,2,3,4,5,6,7,8,10]))
```

```python
#two sum

def twosum(nums, t):

  for i in range(len(nums)):
    for j in range(len(nums)):

      if nums[i] + nums[j] == t:
        return [i, j]

  return "Not found"

twosum([2, 7, 11, 15], 9)
```

```python
#find anagrams

def anagrams(s,t):

  freq1 = {}
  freq2 = {}

  for ch in s:
    freq1[ch] = freq1.get(ch, 0) + 1
  for chs in t:
    freq2[chs] = freq2.get(chs, 0) + 1

  return freq1==freq2

s = "anagram"
t = "nagaram"

anagrams(s,t)
```

```python
#maximum subarray sum kadanes algo

def max_subarray(nums):
  max_sum = nums[0]
  current_sum = nums[0]

  for i in range(1, len(nums)):
    current_sum = max(nums[i], current_sum + nums[i])
    max_sum = max(max_sum, current_sum)

  return max_sum

print(max_subarray([-2, 1, -3, 4, -1, 2, 1, -5, 4]))
```

```python
#rotate by k positions

def rotate(nums, k):
  n = len(nums)
  k = k%n

  nums.reverse()

  nums[:k] = reversed(nums[:k])
  nums[k:] = reversed(nums[k:])

  return nums

nums = [1, 2, 3, 4, 5, 6, 7]

print(rotate(nums, 3))
```

```python
#second largest element

def second_large(nums):

  nums.sort()
  return nums[len(nums) - 2]

print(second_large([2, 1, 3, 4, 5, 6, 7]))
```
