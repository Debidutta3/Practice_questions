```python
def two_sum(num, target):
  for i in range(len(num)):
    for j in range(i+1, len(num)):
      if num[i] + num[j] == target:
        return [i, j]
  return []

print(two_sum([2, 7, 11, 15], 9))
```

```python
def rotate(nums, k):

    k = k%len(nums)

    res = nums[-k:] + nums[:-k]

    return res

nums = [1, 2, 3, 4, 5, 6, 7]
k = 3

print(rotate(nums, k))
```

```python
#maximum subarray sum

def max_subarray(nums):
  curr = nums[0]
  max_s = nums[0]

  for i in range(1, len(nums)):
    curr = max(nums[i], curr+ nums[i])
    max_s = max(max_s, curr)

  return max_s

nums = [-2, 1, -3, 4, -1, 2, 1, -5, 4]

print(max_subarray(nums))
```

```python
#missing number

def miss(nums):
  j = 1

  for i in range(len(nums)):

    if nums[i] != j:
      missing = j
      j+=1

  return missing

nums = [1, 2, 4, 5, 6]
miss(nums)
```

```python
#Count pairs with a given difference

def pairs_k(nums, k):

  count = 0

  for i in nums:
    for j in nums:
      if i-j == k:
        count+=1

  return count

nums = [1, 5, 3, 4, 2]
k = 2

pairs_k(nums, k)
```

```python
#Merge Two Sorted Arrays

def merge_arrays(num1, num2):

  new_l = []

  for i in num1:
    new_l.append(i)

  for j in num2:
    new_l.append(j)

  new_l.sort()

  return new_l

num1 = [1, 3, 5, 7]
num2 = [2, 4, 6, 8]

merge_arrays(num1, num2)
```

```python
#find most repeated element

def repeated(nums):

  freq={}

  for num in nums:
    freq[num] = freq.get(num, 0) + 1

  max_count = 0
  for num,count in freq.items():
    if count > max_count:
      max_count = count
      most_repeated = num

  return most_repeated

nums = [2, 2, 1, 2, 3, 2, 2]
repeated(nums)
```
